```python
total_sales = 0

for i in range(5):
    sales = float(input(f"Sales amount for employee {i + 1}: Rs. "))

    if sales < 50000:
        tax = sales * 0.05
    elif sales <= 100000:
        tax = sales * 0.10
    else:
        tax = sales * 0.15

    total_sales += sales

    print("Sales: Rs.", int(sales))
    print("Tax: Rs.", int(tax))

print("Total Sales: Rs.", int(total_sales))

average_sales = total_sales / 5
print("Average Sales: Rs.", int(average_sales))
```
